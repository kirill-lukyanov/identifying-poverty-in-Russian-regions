# Определение уязвимых групп населения

Задача проекта:
- Кластеризовать регионы России и определить, какие из них наиболее
остро нуждаются в помощи малообеспеченным/неблагополучным
слоям населения;
- Описать группы населения, сталкивающиеся с бедностью;
- Определить:
    - Влияет ли число детей, пенсионеров и других социально уязвимых
    групп на уровень бедности в регионе;
    - Связаны ли уровень бедности/социального неблагополучия с
    производством и потреблением в регионе;
    - Какие ещё зависимости можно наблюдать относительно
    социально незащищённых слоёв населения.

## Данные

### Таблицы

- `cash_real_income_wages_2015_2020` — среднедушевые и реальные денежные доходы населения, номинальная и реальная начисленная зарплата, по регионам.

Папка `crimes` — сведения о преступлениях, совершённых отдельными
категориями лиц за 2016–2022 гг., по месяцам, регионам, категориям лиц,
категориям преступлений:
1. В папке содержатся файлы о расследованных преступлениях (число

пригодный для анализа вид.
3. Данные за каждый месяц агрегированы с предыдущими месяцами
этого года (то есть январь 2022, далее — январь + февраль 2022 и так
далее).
4. Файлы содержат указание на месяц в названии файла по типу
4-EGS_Razdel_%_%MMYYYY%.xls (в названии может быть раздел 4 или 5).
Это связано с тем, что форма статистического наблюдения изменилась
с 2021 года, поэтому файлы различаются, однако по большому срезу
данных в них представлена схожая информация.
Категории лиц, по которым Генпрокуратура представляет статистику:
    - Несовершеннолетние или преступления, совершаемые при их
    соучастии.
    - Ранее совершавшие преступления.
    - Из них ранее судимые.
    - Группа лиц.
    - Группа лиц по предварительному сговору.
    - Организованная группа.
    - Преступное сообщество (преступная организация).
    - В состоянии опьянения.
    - Член семьи, супруг, сожитель, сексуальный партнёр (в том числе
бывший):
        - Из них в отношении женщин.
        - Из них в отношении несовершеннолетних.
    - Иностранные граждане и лица без гражданства:
        - Из них граждан государств-участников СНГ.
        - Из них трудовых мигрантов.
        - Из них незаконных мигрантов.

- Вся информация в сумме приводится на первом листе каждого файла
(«строка 1» в терминологии документов). На остальных строках в
зависимости от года приводится разбивка по категориям преступлений
(особо тяжкие, тяжкие, средней тяжести, небольшой тяжести), а далее,
если есть дополнительные листы (с 2021 года) — по главам и статьям
Уголовного кодекса. Для более ранних годов разбивки по
преступлениям нет — только по категориям преступлений.
- Справка об уголовной ответственности несовершеннолетних. С разного возраста наступает подсудность по разным
статьям. Для одних статей ответственность наступает с 14 лет, для других

```python
# процент людей, живущих за чертой бедности (с денежными доходами ниже величины прожиточного минимума), оценка за год по регионам (процент бедности).																						
poverty_percent = pd.read_csv(data_path+'poverty_percent_by_regions_1992_2020.csv')\
    .rename(columns={'year': 'date'})
poverty_percent['date'] = pd.to_datetime(poverty_percent['date'], format='%Y')
poverty_percent['region'] = poverty_percent['region'].apply(standartize_region)  
poverty_percent.drop_duplicates(inplace=True)
poverty_percent.set_index(['region', 'date'], inplace=True)
poverty_percent
```

```python
# Распределение малоимущего населения по социально-демографическим группам (2017, процент, значение показателя за год)
poverty_socdem_years = [
    '2017',
    '2018',
    '2019',
    '2020',
]

poverty_socdem_dfs = []
for year in poverty_socdem_years:
    df = pd.read_excel(data_path+f'poverty_socdem_{year}.xls', header=2) \
        .rename(columns={
            'Unnamed: 0': 'region',
            'Дети в возрасте до 16 лет': 'below_working_age',
            'Население старше трудоспособного возраста': 'above_working_age',
            'Население трудоспособного возраста': 'of_working_age',
        })
    df.drop(columns=['Все население'], inplace=True)
    df['region'] = df['region'].apply(standartize_region)
    df['date'] = pd.to_datetime(year, format='%Y')
    poverty_socdem_dfs.append(df)

poverty_socdem = pd.concat(poverty_socdem_dfs)
poverty_socdem.set_index(['region', 'date'], inplace=True)
poverty_socdem.columns = ['poverty_'+c for c in poverty_socdem.columns]
# Данные в которых сумма категорий равна нулю, являются пропусками
poverty_socdem.loc[poverty_socdem[poverty_socdem.sum(axis=1)==0].index] = np.nan 
poverty_socdem
```

```python
population_dfs = []
for file_name in os.listdir(data_path+'population'):
    df = pd.read_excel(data_path+'population/'+file_name).iloc[7:103,:5]
    df['date'] = pd.to_datetime(file_name[-18:-14], format='%Y')
    population_dfs.append(df)

population = pd.concat(population_dfs)
population = population.replace('–', np.nan)
population.columns = ['region', 'population', 'population_below_working_age', 'population_of_working_age', 'population_above_working_age', 'date']
population['region'] = population['region'].apply(standartize_region)
population.set_index(['region', 'date'], inplace=True)
# Процент социально-демографических групп
population['population_below_working_age_pct'] = population['population_below_working_age'] / population['population'] * 100
population['population_of_working_age_pct'] = population['population_of_working_age'] / population['population'] * 100
population['population_above_working_age_pct'] = population['population_above_working_age'] / population['population'] * 100
population
```

```python
# Количество малоимущего населения среди социально-демографических групп
poverty_count = population['population'] * poverty_percent['poverty_percent'] / 100
# Процент бедности среди социально-демографических групп
poverty_socdem_ratio = pd.concat([
    (poverty_count * poverty_socdem['poverty_below_working_age'] / 100)
    / population['population_below_working_age'],
    (poverty_count * poverty_socdem['poverty_of_working_age'] / 100)
    / population['population_of_working_age'],
    (poverty_count * poverty_socdem['poverty_above_working_age'] / 100)
    / population['population_above_working_age']
], axis=1).dropna()
poverty_socdem_ratio.columns = ['poverty_ratio_below_working_age',
                                'poverty_ratio_of_working_age', 
                                'poverty_ratio_above_working_age']
poverty_socdem_ratio
```

```python
alco_drugs_sheets = {
    'drugs': [ # Заболеваемость с впервые в жизни установленным диагнозом наркомании на 100 тыс. населения
        pd.read_excel(data_path+'morbidity/drug_alco.xlsx', sheet_name=f'drugs').set_index('region'),
        pd.read_excel(data_path+'morbidity/drug_alco.xlsx', sheet_name=f'drugs1718').set_index('region'),
        pd.read_excel(data_path+'morbidity/drugs_2017_2024.xls', header=2).drop(columns=['2017', '2018']) \
            .drop(index=0, columns='Unnamed: 1').rename(columns={'Unnamed: 0': 'region'}).set_index('region'),
    ],
    'alco': [ # Заболеваемость с впервые в жизни установленным диагнозом алкоголизма и алкогольного психоза на 100 тыс. населения 
        pd.read_excel(data_path+'morbidity/drug_alco.xlsx', sheet_name=f'alco').set_index('region'),
        pd.read_excel(data_path+'morbidity/drug_alco.xlsx', sheet_name=f'alco1718').set_index('region'),
        pd.read_excel(data_path+'morbidity/alco_2018_2024.xls', header=2).drop(columns='2018') \
            .drop(index=0, columns='Unnamed: 1').rename(columns={'Unnamed: 0': 'region'}).set_index('region'),
    ], 
}
alco_drugs_dfs = []

for name, sheets in alco_drugs_sheets.items():
    for i, sheet in enumerate(sheets):
        sheets[i].index = sheet.index.to_series().apply(standartize_region)
    df = reduce(lambda left, right: left.join(right, how='outer'), sheets)
    df = df.drop(index='Главное медицинское управление Управления делами Президента Российской Федерации')
    df.columns = pd.to_datetime(df.columns, format='%Y')
    df = df.astype('float')
    # Для заполнения пропусков в Архангельской и Тюменской областях
    # данные переводятся из относительной величины (заболеваемость на 100 тыс. населения) в абсолютную (количество заболевших),
    # после чего проводится заполнение исходя из вышеописанной структуры и переводятся обратно в относительную величину
    df = (df * population.reset_index().pivot(index='region', columns='date', values='population') / 100000).loc[df.index, df.columns].dropna(thresh=1, axis=1)
    mean_more_5 = df.mean(axis=1) > 5
    df[mean_more_5] = df[mean_more_5].replace(0, np.nan)
    df.loc['Архангельская область'] = df.loc['Архангельская область'].combine_first((df.loc['Архангельская область без НАО'] + df.loc['Ненецкий АО']))
    df.loc['Архангельская область без НАО'] = df.loc['Архангельская область без НАО'].combine_first(df.loc['Архангельская область'] - df.loc['Ненецкий АО'])
    df.loc['Тюменская область'] = df.loc['Тюменская область'].combine_first(df.loc['Тюменская область без ХМАО и ЯНАО']+df.loc['Ханты-Мансийский АО - Югра']+df.loc['Ямало-Ненецкий АО'])
    df.loc['Тюменская область без ХМАО и ЯНАО'] = df.loc['Тюменская область без ХМАО и ЯНАО'].combine_first(df.loc['Тюменская область']-(df.loc['Ханты-Мансийский АО - Югра']+df.loc['Ямало-Ненецкий АО']))
    df = (df * 100000 / population.reset_index().pivot(index='region', columns='date', values='population')).loc[df.index, df.columns].dropna(thresh=1, axis=1)

    regions_to_interpolate = [
        'Архангельская область',
        'Архангельская область без НАО',
        'Астраханская область',
        'Волгоградская область',
        'Дальневосточный федеральный округ',
        'Кабардино-Балкарская Республика',
        'Карачаево-Черкесская Республика',
        'Краснодарский край',
        'Приволжский федеральный округ',
        'Республика Адыгея',
        'Республика Дагестан',
        'Республика Ингушетия',
        'Республика Калмыкия',
        'Республика Крым',
        'Республика Северная Осетия - Алания',
        'Ростовская область',
        'Севастополь',
        'Северо-Западный федеральный округ',
        'Северо-Кавказский федеральный округ',
        'Сибирский федеральный округ',
        'Ставропольский край',
        'Тюменская область',
        'Тюменская область без ХМАО и ЯНАО',
        'Уральский федеральный округ',
        'Центральный федеральный округ',
        'Чеченская Республика',
        'Южный федеральный округ'
    ]

    # Заполнение пропусков временной интерполяцией
    for region in regions_to_interpolate:
        df.loc[region] = df.loc[region].interpolate()

    df = df.melt(value_name=name, var_name='date', ignore_index=False).reset_index()
    alco_drugs_dfs.append(df)

alco_drugs = reduce(lambda left, right: left.merge(right, on=['region', 'date'], how='outer'), alco_drugs_dfs)
alco_drugs.set_index(['region', 'date'], inplace=True)
alco_drugs
```

```python
# Извлечение производственного сектора 
def extract_sector_2005_2016(s: str):
    s = s.strip()
    s = re.sub(r'\s+', ' ', s)
    s = s.replace('"', '').replace("'", '')

    s = str(s).upper()

    if 'ПОДРАЗДЕЛ' in s:
        return 'drop'
    
    if 'РАЗДЕЛ' in s:
        if 'ДОБЫЧА' in s:
            return 'mining'
        elif 'ОБРАБАТЫВАЮЩИЕ' in s:
            return 'manufacturing'
        elif 'ПРОИЗВОДСТВО И РАСПРЕДЕЛЕНИЕ ЭЛЕКТРОЭНЕРГИИ, ГАЗА И ВОДЫ' in s:
            return 'housing_and_utilities'
        else:
            return 'other'
    else:
        return 'drop'
    
def extract_sector_2017_2020(s: str):
    s = s.strip()
    s = re.sub(r'\s+', ' ', s)
    s = s.replace('"', '').replace("'", '')

    # s = str(s).upper()

    if s == 'ДОБЫЧА ПОЛЕЗНЫХ ИСКОПАЕМЫХ':
        return 'mining'
    if s == 'ОБРАБАТЫВАЮЩИЕ ПРОИЗВОДСТВА':
        return 'manufacturing'
    if s == 'Промышленное производство (промышленность)':
        return 'industry'
    if s == 'ОБЕСПЕЧЕНИЕ ЭЛЕКТРИЧЕСКОЙ ЭНЕРГИЕЙ, ГАЗОМ И ПАРОМ; КОНДИЦИОНИРОВАНИЕ ВОЗДУХА':
        return 'housing_and_utilities'
    if s == 'ВОДОСНАБЖЕНИЕ; ВОДООТВЕДЕНИЕ, ОРГАНИЗАЦИЯ СБОРА И УТИЛИЗАЦИИ ОТХОДОВ, ДЕЯТЕЛЬНОСТЬ ПО ЛИКВИДАЦИИ ЗАГРЯЗНЕНИЙ':
        return 'housing_and_utilities'
    else:
        return 'drop'

regional_production_2005_2016 = pd.read_csv(data_path+'regional_production_2005_2016.csv')
regional_production_2005_2016['production_field'] = regional_production_2005_2016['production_field'].apply(extract_sector_2005_2016)
regional_production_2005_2016 = regional_production_2005_2016[regional_production_2005_2016['production_field'] != 'drop']
regional_production_2005_2016['region'] = regional_production_2005_2016['region'].apply(standartize_region)
regional_production_2005_2016 = regional_production_2005_2016.groupby(['region', 'production_field']).max().reset_index()

regional_production_2017_2020 = pd.read_csv(data_path+'regional_production_2017_2020.csv')
regional_production_2017_2020['production_field'] = regional_production_2017_2020['production_field'].apply(extract_sector_2017_2020)
regional_production_2017_2020 = regional_production_2017_2020[regional_production_2017_2020['production_field'] != 'drop']
regional_production_2017_2020 = regional_production_2017_2020.groupby(['region', 'production_field']).sum().reset_index()
regional_production_2017_2020['region'] = regional_production_2017_2020['region'].apply(standartize_region)
regional_production_2017_2020 = regional_production_2017_2020.groupby(['region', 'production_field']).max().reset_index()

regional_production = pd.merge(
    regional_production_2005_2016,
    regional_production_2017_2020, 
    how='outer', on=['region', 'production_field'])

regional_production.set_index(['region', 'production_field'], inplace=True)
regional_production.drop(index=[
    'Москва в старых границах',
    'Московская область в старых границах',
    'Агинский Бурятский округ',
    'Коми-Пермяцкий округ, входящий в состав Пермского края',
    'Корякский округ, входящий в состав Камчатского края',
    'Таймырский (Долгано-Ненецкий) АО (Красноярский край)',
    'Усть-Ордынский Бурятский округ',
    'Эвенкийский АО (Красноярский край)'
], level=0, inplace=True)

regional_production.loc[('Архангельская область без НАО', 'housing_and_utilities')] = regional_production.loc[('Архангельская область', 'housing_and_utilities')] - regional_production.loc[('Ненецкий АО', 'housing_and_utilities')]
regional_production.loc[('Архангельская область без НАО', 'mining')] = regional_production.loc[('Архангельская область', 'mining')] - regional_production.loc[('Ненецкий АО', 'mining')]
regional_production.loc[('Архангельская область без НАО', 'manufacturing')] = regional_production.loc[('Архангельская область', 'manufacturing')] - regional_production.loc[('Ненецкий АО', 'manufacturing')]

regional_production.loc[('Тюменская область без ХМАО и ЯНАО', 'housing_and_utilities')] = regional_production.loc[('Тюменская область', 'housing_and_utilities')] \
    - regional_production.loc[('Ямало-Ненецкий АО', 'housing_and_utilities')] - regional_production.loc[('Ханты-Мансийский АО - Югра', 'housing_and_utilities')]
regional_production.loc[('Тюменская область без ХМАО и ЯНАО', 'mining')] = regional_production.loc[('Тюменская область', 'mining')] \
    - regional_production.loc[('Ямало-Ненецкий АО', 'mining')] - regional_production.loc[('Ханты-Мансийский АО - Югра', 'mining')]
regional_production.loc[('Тюменская область без ХМАО и ЯНАО', 'manufacturing')] = regional_production.loc[('Тюменская область', 'manufacturing')] \
    - regional_production.loc[('Ямало-Ненецкий АО', 'manufacturing')] - regional_production.loc[('Ханты-Мансийский АО - Югра', 'manufacturing')]

regional_production = regional_production.melt(
    var_name='date',
    value_name='production_volume',
    ignore_index=False
).reset_index()

regional_production['date'] = pd.to_datetime(regional_production['date'], format='%Y')
regional_production = regional_production.pivot(index=['region', 'date'], columns='production_field', values='production_volume')
regional_production.columns = regional_production.columns = ['production_'+c for c in regional_production.columns]
regional_production = regional_production.fillna(0)
regional_production
```

```python
def stat_test(data, alpha=0.05, alternative='two-sided', norm=None, verbose=0, title_plt=None, xtitle=None):
    data = pd.DataFrame(data)

    # Проверка выборок на нормальность
    if norm is None:
        norm = True
        for i in range(data.shape[1]):
            if shapiro(data.iloc[:,i]).pvalue <= alpha:
                norm = False
                break

    if data.shape[1] == 2:
        if norm:
            pvalue = ttest_ind(data.iloc[:,0], data.iloc[:,1], alternative=alternative).pvalue
            test_name = 'ttest_ind'
        else:
            pvalue = mannwhitneyu(data.iloc[:,0], data.iloc[:,1], alternative=alternative).pvalue
            test_name = 'mannwhitneyu'
    elif data.shape[1] > 2:
        if norm:
            pvalue = f_oneway(data).pvalue
            test_name = 'f_oneway'
        else:
            pvalue = kruskal(data).pvalue
            test_name = 'kruskal'
    
    if verbose >= 1:
        print(data.agg(['mean', 'median']).to_markdown(tablefmt='github'))            
    if verbose >= 2:
        px.histogram(data, opacity=0.8, width=800, title=title_plt,
                     subtitle=f'test - <b>{test_name}</b>, alternative - <b>{alternative}</b>, norm - <b>{norm}</b>, pvalue=<b>{pvalue:.3f}</b>, result - <b>{'H1' if pvalue <= alpha else 'H0'}</b>') \
            .update_layout(title_x=0.5, margin=dict(t=60, b=10), barmode='overlay', xaxis_title_text=xtitle).show()
    
    return pvalue
```

```python
class OutliersDetector(BaseEstimator, TransformerMixin):
    def __init__(self, random_state):
        self.random_state=random_state
        self.model = None

    def fit(self, X, y=None):
        self.model = IsolationForest(random_state=self.random_state)
        self.model.fit(X)
        self._is_fitted = True 
        return self
    
    def transform(self, X):
        return np.concatenate([X, np.array([self.model.predict(X), self.model.score_samples(X)]).T], axis=1)
    
    def __sklearn_is_fitted__(self):
        return hasattr(self, "_is_fitted") and self._is_fitted
```

```python
preprocessing = Pipeline([
    ('PowerTransformer', PowerTransformer()), # Масштабирование 
    ('OutliersDetector', OutliersDetector(random_state=RANDOM_STATE))
])

X_train = preprocessing.fit_transform(X_train)
X_test = preprocessing.transform(X_test)
```

### Gradient Boosting (LightGBM)

```python
base_params  = {
    'objective': 'mse',
    'metric': 'mse',
    'boosting_type': 'gbdt',
    'verbose': -1,
    'n_jobs': -1,
    'random_state': RANDOM_STATE,
}
```

```python
def objective_reg(trial: optuna.Trial, X_train, y_train, cv, not_tuned_params):
    params = {
        'n_estimators': 10000,
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 1, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 4, 31, step=1),
        'min_samples_leaf': trial.suggest_int('min_samples_leaf', 50, 500, step=1),
        'feature_fraction': trial.suggest_float('feature_fraction', 0.7, 1.0),
        'bagging_fraction': trial.suggest_float('bagging_fraction', 0.7, 1.0),
        'bagging_freq': trial.suggest_int('bagging_freq', 1, 5),
        'max_bin': trial.suggest_int('max_bin', 63, 255),
        'max_depth':  trial.suggest_int('max_depth', 2, 8),
        'lambda_l1': trial.suggest_float('lambda_l1', 4, 10),
        'lambda_l2': trial.suggest_float('lambda_l2', 0.1, 1),
    }

    cv_scores = []
    best_iterations = []

    for train_idx, valid_idx in cv.split(X_train):
        X_train_cv, X_valid_cv = X_train[train_idx], X_train[valid_idx]
        y_train_cv, y_valid_cv = y_train[train_idx], y_train[valid_idx]

        pruning_callback = LightGBMPruningCallback(trial, "l2", valid_name="valid_0")
        model = lgb.LGBMRegressor(**params, **not_tuned_params)
        model.fit(X_train_cv, y_train_cv, eval_set=[(X_valid_cv, y_valid_cv)], eval_metric='l2', eval_names=['valid_0'],
                callbacks=[lgb.early_stopping(stopping_rounds=20, verbose=False), pruning_callback])

        valid_preds = model.predict(X_valid_cv, num_iteration=model.best_iteration_)
        valid_score = metrics.mean_squared_error(y_valid_cv, valid_preds)

        cv_scores.append(valid_score)
        best_iterations.append(model.best_iteration_)

    trial.set_user_attr("n_estimators", int(np.median(best_iterations)))
    return np.mean(valid_score)

study = optuna.create_study(sampler=optuna.samplers.TPESampler(seed=RANDOM_STATE), direction='minimize')
study.optimize(lambda t: objective_reg(t, X_train, y_train, cv, not_tuned_params=base_params), n_trials=100)
```
